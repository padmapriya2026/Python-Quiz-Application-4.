```python

questions = [
    {
        "question": "What is the capital of India?",
        "options": ["Mumbai", "Chennai", "New Delhi", "Kolkata"],
        "answer": "New Delhi"
    },
    {
        "question": "Which planet is known as the Red Planet?",
        "options": ["Earth", "Mars", "Jupiter", "Venus"],
        "answer": "Mars"
    },
    {
        "question": "How many days are there in a week?",
        "options": ["5", "6", "7", "8"],
        "answer": "7"
    },
    {
        "question": "Which language are we using for this project?",
        "options": ["Java", "Python", "C++", "HTML"],
        "answer": "Python"
    },
    {
        "question": "Which animal is known as the King of the Jungle?",
        "options": ["Tiger", "Elephant", "Lion", "Leopard"],
        "answer": "Lion"
    }
]

print("Quiz questions loaded successfully!")
print("Total questions:", len(questions))
```

```python

for number, item in enumerate(questions, start=1):
    print(f"\nQuestion {number}: {item['question']}")

    for option_number, option in enumerate(item["options"], start=1):
        print(f"{option_number}. {option}")

    user_answer = input("Enter your answer (1-4): ")

    print("You selected:", user_answer)

```

```python

score = 0

for number, item in enumerate(questions, start=1):
    print(f"\nQuestion {number}: {item['question']}")

    for option_number, option in enumerate(item["options"], start=1):
        print(f"{option_number}. {option}")

    user_answer = input("Enter your answer (1-4): ")

    if user_answer in ["1", "2", "3", "4"]:
        selected_option = item["options"][int(user_answer) - 1]

        if selected_option == item["answer"]:
            print("Correct answer!")
            score += 1
        else:
            print("Incorrect answer!")
            print("Correct answer:", item["answer"])
    else:
        print("Invalid choice!")

print("\nQuiz completed!")
print("Your final score:", score, "/", len(questions))
```

```python

while True:
    score = 0

    for number, item in enumerate(questions, start=1):
        print(f"\nQuestion {number}: {item['question']}")

        for option_number, option in enumerate(item["options"], start=1):
            print(f"{option_number}. {option}")

        user_answer = input("Enter your answer (1-4): ")

        if user_answer in ["1", "2", "3", "4"]:
            selected_option = item["options"][int(user_answer) - 1]

            if selected_option == item["answer"]:
                print("Correct answer!")
                score += 1
            else:
                print("Incorrect answer!")
                print("Correct answer:", item["answer"])
        else:
            print("Invalid choice!")

    print("\n--- QUIZ RESULT ---")
    print("Your final score:", score, "/", len(questions))

    if score == len(questions):
        print("Excellent! Perfect score!")
    elif score >= 3:
        print("Good job! Keep learning!")
    else:
        print("Keep practicing! You can improve!")

    replay = input("\nDo you want to play again? (yes/no): ").lower()

    if replay != "yes":
        print("Thank you for playing!")
        break
```

# Python Quiz Application

## Project Description
The Python Quiz Application is a beginner-friendly General Knowledge quiz. It displays multiple-choice questions, accepts user answers, checks the answers, calculates the score, and displays the final result with a performance message.

## Features
- Contains 5 multiple-choice questions.
- Displays one question at a time.
- Accepts answers from the user.
- Provides correct or incorrect feedback.
- Calculates and displays the final score.
- Shows a performance message based on the score.
- Allows the user to replay the quiz.

## Technologies Used
- Python
- Google Colab

## Python Concepts Used
- Lists
- Dictionaries
- For loops and while loops
- Conditional statements
- User input and output
- Variables and score calculation

## Questions Format
Each question is stored in a Python dictionary containing:
- `question`: The question text.
- `options`: A list of four answer choices.
- `answer`: The correct answer.

All questions are stored together in a Python list.

## Scoring Rules
- Each correct answer earns 1 point.
- Incorrect answers earn 0 points.
- The maximum score is 5.
- A performance message is displayed based on the final score.

## How to Run
1. Open the project in Google Colab.
2. Run the cell containing the questions list.
3. Run the quiz application code.
4. Enter an option from 1 to 4 for each question.
5. View the final score and choose whether to replay.
